# Geant4 tutorial: analysis macros

This notebook runs the analysis macros of the exercises, e.g. `dEdx.C`.
While a macro runs, the cell shows

* the number of simulated events and the last lines of the macro's output,
* the 3D view of the latest event,
* the histograms the macro fills (updated every 2 s).

The canvases the macro draws at the end appear below the cell.
The complete output is written to a log file, e.g. `dEdx.log`.

**Run one macro per kernel session:**

1. Restart the kernel (*Restart* in the notebook toolbar).
2. Run the setup cell below.
3. Run the cell of the macro you want.

Each macro sets up its own geometry with `app->InitMC(...)`, and Geant4 can be
initialised only once per kernel. So restart the kernel before every macro, and also
after you have changed a macro or a geometry file.

```python
# Setup: run once after every kernel restart.
# The first run compiles the tutorial application and takes longer.
import ROOT

ROOT.gROOT.ProcessLine('.x run_g4.C(false)')  # load Geant4 and the tutorial application
app = ROOT.TutorialApplication.Instance()     # the object the macros call `app`

from eventdisplay import run_macro, print_tracks, show3d
```

## Running a macro

`run_macro('dEdx.C')` runs the macro like `.x dEdx.C` in ROOT and adds the live display.
It accepts these options:

| Call | Effect |
|---|---|
| `run_macro('dEdx.C+')` | A trailing `+` compiles the macro before running it. Larger macros that define their own functions or classes need this. |
| `run_macro('dEdx.C', interval=5)` | Updates the display every 5 s instead of every 2 s, so the macro runs faster. |
| `run_macro('dEdx.C', pmin=0.01)` | The 3D view shows only tracks with momentum above 0.01 GeV (10 MeV). This makes events with many particles, such as showers, easier to read. |

After the macro has finished, `print_tracks()` lists the tracks of the last event
and `show3d()` shows that event again in 3D. Both also take `pmin`.

To change the particle, the momentum, the number of events or the geometry, edit the
parameters at the top of the macro file, save it, restart the kernel, and run the
setup cell and the macro cell again.

## Energy loss: `dEdx.C`

Shoots 1000 positive muons with p = 1 GeV through a 2 mm copper block (`geometry/cubox`)
and records the energy each muon deposits in the block.

At the end you get two canvases:

* `c1`: the distribution of the energy loss.
* `c2`: the mean energy loss per unit length and density, −dE/dx, against βγ.

`c2` has one point per momentum. To see the Bethe-Bloch curve, run over several momenta.

The macro prints βγ and −dE/dx for every event. This text goes to **`dEdx.log`** in this
folder, not into the notebook: the cell shows only the last lines while the macro runs.
Open `dEdx.log` in the editor to see all of it.
