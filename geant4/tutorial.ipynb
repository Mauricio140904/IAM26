# Geant4 tutorial: interactive event display

This notebook replaces the old ROOT GUI. It runs the same `TutorialApplication`
as `root -l run_g4.C`. The code of the event display is in `eventdisplay.py`.

**Restart the kernel** (*Restart* button above) whenever you want to load another
geometry or have changed a geometry file: Geant4 can only be initialised once per session.

```python
import ROOT

# load Geant4, compile the tutorial application, create `app`
ROOT.gROOT.ProcessLine('.x run_g4.C(false)')
app = ROOT.TutorialApplication.Instance()

from eventdisplay import event_gui, print_tracks, show3d
```

## Choose the geometry

e.g. `geometry/pbbox` (20 cm lead), `geometry/cubox` (2 mm copper).

```python
app.InitMC('geometry/pbbox')
```

## Shoot particles

Choose particle type and momentum, then press **Next event**.

**3D view**: rotate with the mouse, zoom with the mouse wheel, and hover over a track
to see the particle type, track id, mother and momentum.
Untick **3D view** for the side view of the detector as a picture.

Track colours: e- red, e+ blue, gamma green, mu± magenta, pi± orange, pi0 light orange, p cyan, n dark grey, others grey.

```python
event_gui(app)
```

## Which particle is which?

`print_tracks(pmin)` lists all tracks of the last event with momentum above `pmin` (GeV):
track id, id of the mother track (-1: primary), particle type, PDG code and the
momentum at the point where the particle was created.

```python
print_tracks(pmin=0.01)
```

The 3D view on its own, e.g. larger or only with tracks above `pmin` (GeV):

```python
show3d(pmin=0.01, width=900, height=900)
```

## Histograms

The application fills `hPrimaryEnergy`, `hEdepLong` and `hEdepTrans`.
They are reset with `app.FinishRun()`.

```python
c2 = ROOT.TCanvas('c2', 'primary energy', 700, 500)
ROOT.gROOT.FindObjectAny('hPrimaryEnergy').Draw()
c2.Draw()
```
